```python
import pandas as pd
import numpy as np  
import matplotlib.pyplot as plt
```

```python

df = pd.read_excel("Project Excel V.2.xlsx", sheet_name='Sheet1')


print(f"   Rows    : {df.shape[0]:,}")
print(f"   Columns : {df.shape[1]}")
```

```python


print("BEFORE:")
print(df.columns.tolist())


df.columns = (
    df.columns
      .str.strip()         
      .str.lower()       
      .str.replace(' ', '_') 
)

print("\nAFTER:")
print(df.columns.tolist())
```

```python
print(df.columns.tolist())
```

```python

print("─── Column Data Types ───\n")
print(df.dtypes)

print("\n─── Summary ───")
print(f"Total columns: {df.shape[1]}")
print(f"Numeric      : {df.select_dtypes(include='number').shape[1]}")
print(f"Datetime     : {df.select_dtypes(include='datetime').shape[1]}")
print(f"Category     : {df.select_dtypes(include='category').shape[1]}")
print(f"Object (text): {df.select_dtypes(include='object').shape[1]}")
```

```python

nulls = df.isnull().sum()
null_pct = (nulls / len(df) * 100).round(2)

null_report = pd.DataFrame({
    'missing_count': nulls,
    'missing_%': null_pct
})

print("─── Missing Values ───")
print(null_report)


dup_count = df.duplicated().sum()
print(f"\n─── Duplicates ───")
print(f"Duplicate rows: {dup_count}")


unique_ids = df['order_id'].nunique()
print(f"\n─── Order ID Check ───")
print(f"Total rows : {len(df):,}")
print(f"Unique IDs : {unique_ids:,}")
print(f"Status     : {' All unique' if unique_ids == len(df) else ' Some duplicates'}")
```

```python

df['year']    = df['order_date'].dt.year
df['month']   = df['order_date'].dt.month
df['quarter'] = df['order_date'].dt.quarter 

df['delivery_days'] = (df['ship_date'] - df['order_date']).dt.days


df['profit_margin'] = np.where(
    df['total_revenue'] > 0,
    (df['total_profit'] / df['total_revenue'] * 100).round(2),
    0
)

new_cols = ['order_date', 'ship_date', 'year', 'month',
            'quarter', 'delivery_days', 'profit_margin']
df[new_cols].head()
```

```python


print("─── Shape ───")
print(f"{df.shape[0]:,} rows × {df.shape[1]} columns\n")

print("─── Data Types ───")
print(df.dtypes, "\n")

print("─── Date Range ───")
print(f"From : {df['order_date'].min().date()}")
print(f"To   : {df['order_date'].max().date()}")
print(f"Span : {df['year'].nunique()} years\n")

print("─── Categories ───")
print(f"Regions       : {df['region'].nunique()}")
print(f"Countries     : {df['country'].nunique()}")
print(f"Item Types    : {df['item_type'].nunique()}")
print(f"Sales Channels: {df['sales_channel'].nunique()}\n")

print("─── Key Numbers ───")
print(f"Total Revenue : ${df['total_revenue'].sum():,.0f}")
print(f"Total Profit  : ${df['total_profit'].sum():,.0f}")
print(f"Avg Margin    : {df['profit_margin'].mean():.2f}%")
print(f"Avg Delivery  : {df['delivery_days'].mean():.1f} days")
```

```python


pivot_region = df.pivot_table(
    index='region',
    values=['total_revenue', 'total_profit', 'units_sold'],
    aggfunc='sum'
).round(0).sort_values('total_profit', ascending=False)


pivot_region['margin_%'] = (
    pivot_region['total_profit'] / pivot_region['total_revenue'] * 100
).round(2)

pivot_region = pivot_region[['units_sold', 'total_revenue', 'total_profit', 'margin_%']]

print("─── Performance by Region ───")
pivot_region
```

```python


pivot_item = df.pivot_table(
    index='item_type',
    values=['total_revenue', 'total_profit', 'units_sold'],
    aggfunc='sum'
).round(0).sort_values('total_profit', ascending=False)

pivot_item['margin_%'] = (
    pivot_item['total_profit'] / pivot_item['total_revenue'] * 100
).round(2)

pivot_item = pivot_item[['units_sold', 'total_revenue', 'total_profit', 'margin_%']]

print("─── Performance by Item Type ───")
pivot_item
```

```python


pivot_channel = df.pivot_table(
    index='sales_channel',
    values=['total_revenue', 'total_profit', 'units_sold', 'delivery_days'],
    aggfunc={'total_revenue': 'sum',
             'total_profit':  'sum',
             'units_sold':    'sum',
             'delivery_days': 'mean'}
).round(2)

print("─── Online vs Offline ───")
pivot_channel
```

```python

pivot_year = df.pivot_table(
    index='region',
    columns='year',
    values='total_profit',
    aggfunc='sum'
).round(0)


pivot_year['Total'] = pivot_year.sum(axis=1)
pivot_year = pivot_year.sort_values('Total', ascending=False)

print("─── Profit by Region across Years ───")
pivot_year
```

```python

pivot_country = df.pivot_table(
    index='country',
    values=['total_revenue', 'total_profit', 'units_sold'],
    aggfunc='sum'
).round(0).sort_values('total_profit', ascending=False).head(10)

pivot_country['margin_%'] = (
    pivot_country['total_profit'] / pivot_country['total_revenue'] * 100
).round(2)

pivot_country = pivot_country[['units_sold', 'total_revenue', 'total_profit', 'margin_%']]

print("─── Top 10 Countries by Profit ───")
pivot_country
```

```python
data = pivot_region['total_profit'].sort_values()

plt.figure(figsize=(10, 6))
plt.barh(data.index, data.values, color='#0969da')
plt.title("Profit by Region", fontsize=13, fontweight='bold')
plt.xlabel("Total Profit ($)")
plt.tight_layout()
plt.show();
```

```python

data = pivot_item['total_profit']

plt.figure(figsize=(11, 6))
plt.bar(data.index, data.values, color='#1a7f37')
plt.title("Profit by Item Type", fontsize=13, fontweight='bold')
plt.ylabel("Total Profit ($)")
plt.xticks(rotation=40, ha='right')
plt.tight_layout()
plt.show();
```

```python

years_only = pivot_year.drop(columns='Total')

fig, ax = plt.subplots(figsize=(13, 6))

colors = ['#0969da', '#1a7f37', '#bc4c00', '#8250df',
          '#cf222e', '#1b7c83', '#9a6700']

for i, region in enumerate(years_only.index):
    series = years_only.loc[region].dropna()         
    x = [int(y) for y in series.index]               
    y = series.values / 1e6                          

    ax.plot(x, y,
            marker='o', markersize=8, linewidth=2.5,
            color=colors[i % len(colors)], label=region)


ax.set_title("Yearly Profit Trend by Region", fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel("Year", fontsize=11)
ax.set_ylabel("Total Profit ($ Millions)", fontsize=11)

all_years = [int(y) for y in years_only.columns]
ax.set_xticks(all_years)

ax.grid(True, linestyle='--', alpha=0.5)
ax.set_axisbelow(True)


ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left',
          frameon=False, fontsize=10)


ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

plt.tight_layout()
plt.show();
```
